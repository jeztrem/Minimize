```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

```python

x_grid = np.linspace(-2, 2, 500)
a = 0
step_size = 0.05
starting_position = 0.0

def potential(x):
    return (x[0]**2 - 1)**2 + a*x[0]

def gradient(x):
    return np.array([
        4*x[0]*(x[0]**2 - 1) + a
    ])
```

```python
def steepest_descent(f, grad, x0, step_size, tol=1e-8, max_iter=5000):
    x = np.array(x0,dtype=float)
    path = [x.copy()]
    for k in range(max_iter):
        g = grad(x)
        if np.linalg.norm(g) < tol:
            break
        x = x - step_size*g
        path.append(x.copy())
    return {"x":x,"f":f(x),"iterations":k,"path":np.array(path)}
```

```python
start=[starting_position]
sd_result = steepest_descent(potential, gradient, start, step_size)
current_history = sd_result["path"][:, 0]
best_x = current_history[0]
best_energy = float(potential([best_x]))
best_history = []

for current_x in current_history:
    current_energy = float(potential([current_x]))
    if current_energy < best_energy:
        best_x = current_x
        best_energy = current_energy
    best_history.append(best_x)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x_grid, [potential([xi]) for xi in x_grid], color="black", linewidth=2)
current_point, = ax.plot([], [], "o", color="red", markersize=12)
best_point, = ax.plot([], [], "o", color="green", markeredgecolor="black", markerfacecolor="none", markersize=14)
ax.set(xlabel="x", ylabel="V(x)", title="Steepest descent")


def update(frame):
    current = current_history[frame]
    best = best_history[frame]
    current_point.set_data([current], [potential([current])])
    best_point.set_data([best], [potential([best])])
    return current_point, best_point

animation = FuncAnimation(
    fig, update, frames=len(current_history), interval=100, blit=True, repeat=False
)
plt.close(fig)
HTML(animation.to_jshtml())
```
