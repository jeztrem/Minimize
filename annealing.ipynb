```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation
from IPython.display import HTML
```

```python
x_grid = np.linspace(-2, 2, 500)
a = 0
starting_position = 0.0
starting_temperature = 5.0
step_size = 2.5
n_steps = 200


def potential(x):
    return (x[0]**2 - 1)**2 + a*x[0]

def gradient(x):
    return np.array([
        4*x[0]*(x[0]**2 - 1) + a
    ])
```

```python




rng = np.random.default_rng(7)
temperatures = np.geomspace(starting_temperature, 0.02, n_steps)
current_x = starting_position
current_energy = float(potential([current_x]))
best_x = current_x
best_energy = current_energy
current_history = [current_x]
best_history = [best_x]

for temperature in temperatures:
    proposal_x = np.clip(current_x + rng.normal(0, step_size), -2, 2)
    proposal_energy = float(potential([proposal_x]))
    accepted = proposal_energy <= current_energy or rng.random() < np.exp(
        -(proposal_energy - current_energy) / temperature
    )
    if accepted:
        current_x = proposal_x
        current_energy = proposal_energy
        if current_energy < best_energy:
            best_x = current_x
            best_energy = current_energy
    current_history.append(current_x)
    best_history.append(best_x)

fig, ax = plt.subplots(figsize=(10, 5))
ax.plot(x_grid, [potential([xi]) for xi in x_grid], color="black", linewidth=2)
current_point, = ax.plot([], [], "o", color="red", markersize=12)
best_point, = ax.plot([], [], "o", markeredgecolor="black", markerfacecolor="none", markersize=14)
ax.set(xlabel="x", ylabel="V(x)", title="Simulated Annealing")


def update(frame):
    current = current_history[frame]
    best = best_history[frame]
    current_point.set_data([current], [potential([current])])
    best_point.set_data([best], [potential([best])])
    return current_point, best_point

animation = FuncAnimation(
    fig, update, frames=n_steps + 1, interval=100, blit=True, repeat=False
)
plt.close(fig)
HTML(animation.to_jshtml())
```
